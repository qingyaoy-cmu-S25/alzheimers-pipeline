## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load Sample Data

```python
# Create sample research data
np.random.seed(42)
n_samples = 100

data = pd.DataFrame({
    'sample_id': [f'S{i:03d}' for i in range(n_samples)],
    'gene_expression': np.random.normal(5, 2, n_samples),
    'protein_level': np.random.normal(100, 30, n_samples),
    'age': np.random.randint(40, 90, n_samples),
    'group': np.random.choice(['Control', 'Treatment'], n_samples)
})

print(f"Dataset shape: {data.shape}")
data.head(10)
```

## 3. Data Summary Statistics

```python
# Summary statistics by group
summary = data.groupby('group').agg({
    'gene_expression': ['mean', 'std'],
    'protein_level': ['mean', 'std'],
    'age': ['mean', 'min', 'max']
}).round(2)

print("Summary Statistics by Group:")
summary
```

## 4. Visualize Gene Expression Distribution

```python
# Create histogram of gene expression by group
fig, ax = plt.subplots(figsize=(10, 6))

for group in ['Control', 'Treatment']:
    group_data = data[data['group'] == group]['gene_expression']
    ax.hist(group_data, bins=20, alpha=0.6, label=group)

ax.set_xlabel('Gene Expression Level')
ax.set_ylabel('Frequency')
ax.set_title('Gene Expression Distribution by Group')
ax.legend()
plt.tight_layout()
plt.show()
```

## 5. Scatter Plot: Gene Expression vs Protein Level

```python
# Create scatter plot
fig, ax = plt.subplots(figsize=(10, 6))

colors = {'Control': '#3b82f6', 'Treatment': '#ef4444'}
for group in ['Control', 'Treatment']:
    group_data = data[data['group'] == group]
    ax.scatter(group_data['gene_expression'], group_data['protein_level'], 
               c=colors[group], label=group, alpha=0.6, s=50)

ax.set_xlabel('Gene Expression')
ax.set_ylabel('Protein Level')
ax.set_title('Gene Expression vs Protein Level')
ax.legend()
plt.tight_layout()
plt.show()
```

## 6. Statistical Analysis

```python
from scipy import stats

# T-test between groups
control = data[data['group'] == 'Control']['gene_expression']
treatment = data[data['group'] == 'Treatment']['gene_expression']

t_stat, p_value = stats.ttest_ind(control, treatment)

print("T-Test Results (Gene Expression):")
print(f"  Control mean: {control.mean():.3f}")
print(f"  Treatment mean: {treatment.mean():.3f}")
print(f"  T-statistic: {t_stat:.3f}")
print(f"  P-value: {p_value:.4f}")

if p_value < 0.05:
    print("  Result: Statistically significant difference (p < 0.05)")
else:
    print("  Result: No significant difference (p >= 0.05)")
```
